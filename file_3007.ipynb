```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
```

```python
df = pd.DataFrame({
    'rollNo' : [1, 2, 3, None, 5, None, 7, None, 9, 10],
    'Name' : ['Harsh', 'Idris', 'Raza', 'Vardhan', 'Santosh', 'Kumar', None, 'Shashi', 'Manjuth', None],
    'marks' : [35, 50, None, 89, 90, 99, None, 100, None, None ],
    'Grade' : ['C', 'B', np.nan, 'A', 'A+', 'A+', np.nan, 'A+', np.nan, np.nan ]

})
```

```python
df
```

```python
imputer = SimpleImputer(strategy='constant', fill_value='unknown')
imputer.fit_transform(df[['Name', 'Grade']])
```

```python
imputer = SimpleImputer(strategy='mean')
imputer.fit_transform(df[['rollNo', 'marks']])
```

```python
imputer = SimpleImputer(strategy='median')
imputer.fit_transform(df[['rollNo', 'marks']])
```

```python
imputer = SimpleImputer(strategy='constant', fill_value=0)
imputer.fit_transform(df[['rollNo', 'marks']])
```

```python
imputer = SimpleImputer(strategy='constant', fill_value='unknown')
imputer.fit_transform(df[['Name', 'Grade']])
```
