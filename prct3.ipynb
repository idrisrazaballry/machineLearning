```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
df = pd.read_csv('dataset_3.csv')
df
```

```python
df.info()
```

```python
df.describe()
```

```python
x = df.drop(columns=['is_fraud'])
y = df['is_fraud']
```

```python
xtrain, xtest, ytrain, ytest = train_test_split(x, y, random_state=42, train_size=0.8, stratify=y)
```

```python
y.value_counts()
```

```python
scaler = StandardScaler()
scaler.fit_transform(xtrain)
scaler.transform(xtest)
```

```python
logisticModel = LogisticRegression(max_iter=1000, class_weight='balanced')

logisticModel.fit(xtrain, ytrain)
```

```python
y_pred = logisticModel.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score, confusion_matrix

AccurayScore = accuracy_score(ytest, y_pred)
AccurayScore
```

```python
RecallScore = recall_score(ytest, y_pred)
RecallScore
```

```python
PrecisionScore = precision_score(ytest, y_pred)
PrecisionScore
```

```python
F1Score = f1_score(ytest, y_pred)
F1Score
```

```python
ConfusionMatrix = confusion_matrix(ytest, y_pred)
ConfusionMatrix
```
